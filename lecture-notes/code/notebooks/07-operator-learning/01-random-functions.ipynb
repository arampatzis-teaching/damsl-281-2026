# Random functions

Section 7.4 of the notes.

Operator learning draws its inputs from a distribution over functions. This
notebook draws functions from the distribution of Section 7.4, a random sum of
the first 16 waves on $(-1,1)$, and plots a few of them.

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

FIGURES = Path("figures")
FIGURES.mkdir(exist_ok=True)
```

## 1. The recipe

$u(x) = \frac{A}{Z}\sum_{k=1}^{16} c_k\,(\alpha_k\cos\pi kx + \alpha_k'\sin\pi kx)$
with $c_k = 1/(1+k^2/16)$, $Z = (\sum_k c_k^2)^{1/2}$, $A = 0.5$, and
$\alpha_k$, $\alpha_k'$ independent standard Gaussians. Drawing a function means
drawing the 32 numbers $\alpha_k$, $\alpha_k'$.

```python
def random_functions(x, n_samples, n_waves, A, rng):
    # Rows: n_samples random sums of the first n_waves waves, evaluated at x.
    k = np.arange(1, n_waves + 1)
    c = 1 / (1 + k**2 / 16)
    Z = np.sqrt((c**2).sum())
    alpha = rng.standard_normal((n_samples, n_waves))
    beta = rng.standard_normal((n_samples, n_waves))
    phase = np.pi * np.outer(k, x)
    return A / Z * ((alpha * c) @ np.cos(phase) + (beta * c) @ np.sin(phase))


x = np.linspace(-1, 1, 512)
u = random_functions(x, n_samples=5, n_waves=16, A=0.5, rng=np.random.default_rng(1))
big = random_functions(
    x, n_samples=20000, n_waves=16, A=0.5, rng=np.random.default_rng(2)
)
print(f"mean over draws {big.mean():.3f}, standard deviation {big.std():.3f}")
```

```python
plt.rcParams.update(
    {
        "font.family": "STIXGeneral",
        "mathtext.fontset": "stix",
        "font.size": 8,
        "axes.spines.top": False,
        "axes.spines.right": False,
    }
)
colors = ["#2a78d6", "#eb6834", "#1baf7a", "#8e3fb8", "#8a8983"]
fig, ax = plt.subplots(figsize=(3.0, 1.0))
for row, color in zip(u, colors):
    ax.plot(x, row, color=color, lw=1.1)
ax.set(xlabel="$x$", ylabel="$u(x)$", xlim=(-1, 1), xticks=[-1, -0.5, 0, 0.5, 1])
fig.savefig(FIGURES / "random_functions.pdf", bbox_inches="tight")
plt.show()
```
