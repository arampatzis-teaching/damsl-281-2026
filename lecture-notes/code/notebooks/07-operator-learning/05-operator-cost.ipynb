# When a trained operator pays off

Total time to answer $n$ inputs: the solver pays $7.3$ milliseconds per input; the
Fourier neural operator pays once for its data ($1200$ solves) and its training
($16.4$ seconds), then $48$ microseconds per input. The times are those measured in
Section 7.4 of the notes (simulations/notebooks/operators.ipynb, section 9) (solver on one processor core, FNO on the graphics processor).

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

FIGURES = Path("figures")
FIGURES.mkdir(exist_ok=True)
plt.rcParams.update(
    {
        "font.family": "STIXGeneral",
        "mathtext.fontset": "stix",
        "font.size": 8,
        "axes.titlesize": 8,
        "axes.spines.top": False,
        "axes.spines.right": False,
    }
)
```

```python
def total_times(n, t_solve, n_data, t_train, t_eval):
    # Total time for n inputs: solver alone, and data + training + evaluations of the model.
    return n * t_solve, n_data * t_solve + t_train + n * t_eval


t_solve, n_data, t_train, t_eval = 7.30e-3, 1200, 16.4, 48.3e-6
offline = n_data * t_solve + t_train
n_even = offline / (t_solve - t_eval)
print(f"offline {offline:.1f} s, break-even at n = {n_even:.0f}")
```

```python
n = np.logspace(1, 6, 200)
solver, fno = total_times(n, t_solve, n_data, t_train, t_eval)
fig, ax = plt.subplots(figsize=(3.2, 1.6))
ax.loglog(n, solver, color="black", lw=1.3, label="solver")
ax.loglog(n, fno, color="#eb6834", lw=1.3, label="FNO: data, training, evaluations")
ax.axvline(n_even, color="0.6", lw=0.8, ls="--")
ax.text(
    n_even * 1.15, 0.3, f"$n \\approx {round(n_even, -2):.0f}$", color="0.4", fontsize=7
)
ax.set(xlabel="number of inputs $n$", ylabel="total time (s)")
ax.legend(frameon=False, fontsize=7, loc="upper left")
fig.savefig(FIGURES / "operator_cost.pdf", bbox_inches="tight")
plt.show()
```
