# Fourier series: waves and partial sums

Figures for the slides of Chapter 7: the waves $\cos\pi kx$ and $\sin\pi kx$ on $(-1,1)$, and partial sums of the Fourier series of the initial condition $u_0(x) = x^2\cos\pi x$ of the running example.

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

FIGURES = Path("figures")
FIGURES.mkdir(exist_ok=True)
plt.rcParams.update(
    {
        "font.family": "STIXGeneral",
        "mathtext.fontset": "stix",
        "font.size": 8,
        "axes.titlesize": 8,
        "axes.spines.top": False,
        "axes.spines.right": False,
    }
)
```

## The waves

The wave of frequency $k$ has $k$ periods on $(-1,1)$.

```python
x = np.linspace(-1, 1, 801)
fig, axes = plt.subplots(
    1, 3, figsize=(5.2, 1.05), sharey=True, gridspec_kw=dict(wspace=0.2)
)
for ax, k in zip(axes, [1, 2, 4]):
    ax.plot(x, np.cos(np.pi * k * x), color="#2a78d6", lw=1.1, label=r"$\cos\pi kx$")
    ax.plot(
        x,
        np.sin(np.pi * k * x),
        color="#eb6834",
        lw=1.1,
        ls="--",
        label=r"$\sin\pi kx$",
    )
    ax.set(xlabel="$x$", title=f"$k = {k}$", xticks=[-1, 0, 1], yticks=[-1, 0, 1])
fig.legend(
    *axes[0].get_legend_handles_labels(),
    loc="lower center",
    bbox_to_anchor=(0.5, 1.0),
    ncol=2,
    frameon=False,
    fontsize=7,
    handlelength=1.6,
)
fig.savefig(FIGURES / "fourier_basis.pdf", bbox_inches="tight")
plt.show()
```

## Partial sums

The partial sum $s_N$ keeps the waves $k\le N$. On a fine uniform grid the coefficients come from the fast Fourier transform; setting the others to zero and transforming back gives $s_N$ at the grid points.

```python
def partial_sum(v, n_max):
    # Grid values of the partial sum with the waves k <= n_max, from the grid values v.
    v_hat = np.fft.rfft(v)
    v_hat[n_max + 1 :] = 0
    return np.fft.irfft(v_hat, n=v.size)


d = 2048
xg = -1 + 2 * np.arange(d) / d
u0 = xg**2 * np.cos(np.pi * xg)
fig, axes = plt.subplots(
    1, 3, figsize=(5.2, 1.05), sharey=True, gridspec_kw=dict(wspace=0.2)
)
for ax, n_max in zip(axes, [1, 3, 10]):
    ax.plot(xg, u0, color="0.6", lw=2.2, label="$u_0$")
    ax.plot(xg, partial_sum(u0, n_max), color="#2a78d6", lw=1.0, label="sum of waves")
    ax.set(xlabel="$x$", title=f"waves $k\\leq {n_max}$", xticks=[-1, 0, 1])
fig.legend(
    *axes[0].get_legend_handles_labels(),
    loc="lower center",
    bbox_to_anchor=(0.5, 1.0),
    ncol=2,
    frameon=False,
    fontsize=7,
    handlelength=1.6,
)
fig.savefig(FIGURES / "fourier_partial.pdf", bbox_inches="tight")
plt.show()
```
